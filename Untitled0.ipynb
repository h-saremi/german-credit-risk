```python
import pandas as pd

df = pd.read_csv("german_credit.csv", sep=None, engine="python")
df.columns = df.columns.str.strip()   # فاصله‌های اضافه را حذف می‌کند

print(df.shape)
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import pandas as pd

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

df.groupby("account_check_status")["default"].mean().sort_values().plot.barh(ax=ax[0])
ax[0].set_title("Default rate by checking account status")

bins = pd.cut(df["duration_in_month"], [0, 12, 24, 36, 72])
df.groupby(bins, observed=True)["default"].mean().plot.bar(ax=ax[1])
ax[1].set_title("Default rate by loan duration (months)")

plt.tight_layout()
plt.show()
```

```python
cols = [
    "default", "account_check_status", "duration_in_month", "credit_history",
    "purpose", "credit_amount", "savings", "present_emp_since",
    "installment_as_income_perc", "personal_status_sex", "other_debtors",
    "present_res_since", "property", "age", "other_installment_plans",
    "housing", "credits_this_bank", "job", "people_under_maintenance",
    "telephone", "foreign_worker"
]
df.columns = cols
```

```python
print(df["default"].value_counts(normalize=True))
print(df.describe())
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
df["credit_amount"].plot.hist(bins=30, ax=ax[0], title="credit_amount")
np.log(df["credit_amount"]).plot.hist(bins=30, ax=ax[1], title="log(credit_amount)")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

df.groupby("account_check_status")["default"].mean().sort_values().plot.barh(ax=ax[0])
ax[0].set_title("Default rate by checking account status")

bins = pd.cut(df["duration_in_month"], [0, 12, 24, 36, 72])
df.groupby(bins, observed=True)["default"].mean().plot.bar(ax=ax[1])
ax[1].set_title("Default rate by loan duration (months)")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, RocCurveDisplay

X = df.drop(columns="default")
y = df["default"]

num_cols = X.select_dtypes(include="number").columns
cat_cols = X.select_dtypes(exclude="number").columns

pre = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

model = Pipeline([("pre", pre), ("clf", LogisticRegression(max_iter=1000))])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
model.fit(X_train, y_train)

proba = model.predict_proba(X_test)[:, 1]
print("ROC AUC:", round(roc_auc_score(y_test, proba), 3))
RocCurveDisplay.from_predictions(y_test, proba)
plt.show()
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix

def total_cost(th):
    pred = (proba >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    return 5 * fn + 1 * fp   # وام به فرد بد ۵ برابر گران‌تر است

ths = np.linspace(0.05, 0.95, 91)
costs = [total_cost(t) for t in ths]
best = ths[int(np.argmin(costs))]

print("best threshold:", round(best, 2), "| cost:", min(costs))
print("cost at threshold 0.5:", total_cost(0.5))
```

```python
names = model.named_steps["pre"].get_feature_names_out()
coefs = pd.Series(model.named_steps["clf"].coef_[0], index=names).sort_values()

print("Lower default risk:")
print(coefs.head(8))
print("\nHigher default risk:")
print(coefs.tail(8))
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=cv, scoring="roc_auc")

print("AUC per fold:", scores.round(3))
print("mean:", scores.mean().round(3), "| std:", scores.std().round(3))
```

```python
from sklearn.model_selection import cross_val_predict, StratifiedKFold
import numpy as np
from sklearn.metrics import confusion_matrix

def cost_on(y_true, p, th):
    pred = (p >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred).ravel()
    return 5 * fn + fp

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof = cross_val_predict(model, X_train, y_train, cv=cv, method="predict_proba")[:, 1]

ths = np.linspace(0.05, 0.95, 91)
best_th = ths[int(np.argmin([cost_on(y_train, oof, t) for t in ths]))]

print("threshold chosen on training folds:", round(best_th, 2))
print("test cost at that threshold:", cost_on(y_test, proba, best_th))
print("test cost at 0.5:", cost_on(y_test, proba, 0.5))
print("reject-all:", int((y_test == 0).sum()), "| approve-all:", int(5 * (y_test == 1).sum()))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, confusion_matrix

th = 0.25                                    # یا best_th از اعتبارسنجی متقابل
sens = ["personal_status_sex", "age", "foreign_worker"]

# مدل بدون ویژگی‌های حساس
Xtr = X_train.drop(columns=sens)
Xte = X_test.drop(columns=sens)
num2 = Xtr.select_dtypes(include="number").columns
cat2 = Xtr.select_dtypes(exclude="number").columns

pre2 = ColumnTransformer([
    ("num", StandardScaler(), num2),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat2),
])
model_fair = Pipeline([("pre", pre2), ("clf", LogisticRegression(max_iter=1000))])
model_fair.fit(Xtr, y_train)
proba_fair = model_fair.predict_proba(Xte)[:, 1]

def cost(y_true, p, th):
    pred = (p >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred).ravel()
    return 5 * fn + fp

print("AUC  full:", round(roc_auc_score(y_test, proba), 3),
      "| without sensitive:", round(roc_auc_score(y_test, proba_fair), 3))
print("Cost full:", cost(y_test, proba, th),
      "| without sensitive:", cost(y_test, proba_fair, th))

# نرخ رد شدن در هر گروه
t = X_test.copy()
t["y"] = y_test.values
t["reject_full"] = (proba >= th)
t["reject_fair"] = (proba_fair >= th)
t["age_group"] = np.where(t["age"] < 25, "<25", ">=25")

for g in ["foreign_worker", "age_group"]:
    print("\n", g)
    print(t.groupby(g)[["y", "reject_full", "reject_fair"]].agg(["mean"]).round(2))
    print(t[g].value_counts())
```

```python
good = t[t["y"] == 0]
bad  = t[t["y"] == 1]

print("Good applicants wrongly rejected (false positive rate):")
print(good.groupby("age_group")[["reject_full", "reject_fair"]].mean().round(2))
print(good["age_group"].value_counts())

print("\nBad applicants wrongly approved (miss rate):")
print((1 - bad.groupby("age_group")[["reject_full", "reject_fair"]].mean()).round(2))
print(bad["age_group"].value_counts())
```
